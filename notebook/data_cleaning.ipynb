```python
# loading data(nav_history_cleaning)
import pandas as pd

nav_history = pd.read_csv("../data/raw/02_nav_history.csv")

nav_history.head()
```

```python
nav_history["date"] = pd.to_datetime(nav_history["date"])
nav_history["date"]
```

```python
nav_history.info()
```

```python
nav_history = nav_history.sort_values(["amfi_code", "date"])
nav_history
```

```python
nav_history = nav_history.drop_duplicates()
```

```python
invalid_nav = nav_history[nav_history["nav"] <= 0]

print(len(invalid_nav))
```

```python
nav_history["nav"] = (nav_history.groupby("amfi_code")["nav"].ffill())
nav_history["nav"]
```

```python
nav_history["nav"].isnull().sum()  #no null values
```

```python
# saves cleaned nav_history data to processed folder
nav_history.to_csv("../data/processed/nav_history_clean.csv",index=False)
```

```python
# designing sqllite star scheme 
import pandas as pd
from sqlalchemy import create_engine
```

```python
# Creating database

engine = create_engine("sqlite:///bluestock_mf.db")

print("Database Created")
```

```python
# loading cleaned data into database


nav_history = pd.read_csv("../data/processed/nav_history_clean.csv")

transactions = pd.read_csv("../data/processed/investor_transactions_clean.csv")

performance = pd.read_csv("../data/processed/scheme_performance_clean.csv")


```

```python
fund_master = pd.read_csv("../data/raw/01_fund_master.csv")

aum = pd.read_csv("../data/raw/03_aum_by_fund_house.csv")
```

```python
fund_master.to_sql(
    "dim_fund",
    engine,
    if_exists="replace",
    index=False
)

print("dim_fund loaded")
```

```python
nav_history.to_sql(
    "fact_nav",
    engine,
    if_exists="replace",
    index=False
)

print("fact_nav loaded")
```

```python
transactions.to_sql(
    "fact_transactions",
    engine,
    if_exists="replace",
    index=False
)

print("fact_transactions loaded")
```

```python
aum.to_sql(
    "fact_aum",
    engine,
    if_exists="replace",
    index=False
)

print("fact_aum loaded")
```

```python
# verifying tables
pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table';",
    engine
)
```

```python
# verifying row counts 
print("Fund Master:", len(fund_master))
print("NAV History:", len(nav_history))
print("Transactions:", len(transactions))
print("Performance:", len(performance))
print("AUM:", len(aum))
```

```python
# Top 5 Funds by AUM

query = """
SELECT scheme_name,
       aum_crore
FROM fact_performance
ORDER BY aum_crore DESC
LIMIT 5;
"""

result = pd.read_sql(query, engine)

result
```

```python
# Average NAV Per Month

query = """
SELECT strftime('%Y-%m', date) AS month,
ROUND(AVG(nav),2) AS avg_nav
FROM fact_nav
GROUP BY month
ORDER BY month;
"""

result = pd.read_sql(query, engine)

result
```

```python
monthly_sip = pd.read_csv("../data/raw/04_monthly_sip_inflows.csv")

# Note: for YOY growth we need to load monthly_sip_inflows.csv data and then we can calculate the YoY growth.
```

```python
monthly_sip.to_sql(
    "monthly_sip_inflows",
    engine,
    if_exists="replace",
    index=False
)

print("monthly_sip_inflows loaded successfully")
```

```python
# SIP YoY Growth

query = """
SELECT month,
       sip_inflow_crore,
       yoy_growth_pct
FROM monthly_sip_inflows
ORDER BY month;
"""

pd.read_sql(query, engine)
```

```python
# 1.YoY growth values are unavailable for 2022 because prior-year comparison data is not present.
# 2.Positive YoY growth from 2023 onwards indicates increasing investor participation in SIP investments.
```

```python
# Transactions by State

query = """
SELECT state,
       COUNT(*) AS total_transactions
FROM fact_transactions
GROUP BY state
ORDER BY total_transactions DESC;
"""

pd.read_sql(query, engine)
```

```python
#  funds with expense_ratio < 1%, and 5 

query = """
SELECT scheme_name,
       expense_ratio_pct
FROM fact_performance
WHERE expense_ratio_pct < 1
ORDER BY expense_ratio_pct;
"""

pd.read_sql(query, engine)
```

```python
# Top 5 Funds by 5-Year Return

query = """
SELECT scheme_name,
       return_5yr_pct
FROM fact_performance
ORDER BY return_5yr_pct DESC
LIMIT 5;
"""

pd.read_sql(query, engine)
```

```python
# Average Return by Category

query = """
SELECT category,
       ROUND(AVG(return_3yr_pct),2) AS avg_return
FROM fact_performance
GROUP BY category
ORDER BY avg_return DESC;
"""

pd.read_sql(query, engine)
```

```python
# Funds by Risk Grade

query = """
SELECT risk_grade,
COUNT(*) AS total_funds
FROM fact_performance
GROUP BY risk_grade;
"""

pd.read_sql(query, engine)
```

```python
# Transaction type

query = """
SELECT transaction_type,
ROUND(SUM(amount_inr),2) AS total_amount
FROM fact_transactions
GROUP BY transaction_type;
"""

pd.read_sql(query, engine)
```

```python
# Average transactions by statewise

query = """
SELECT state,
ROUND(AVG(amount_inr),2) AS avg_transaction
FROM fact_transactions
GROUP BY state
ORDER BY avg_transaction DESC;
"""

pd.read_sql(query, engine)
```
