```python
# loading data(nav_history_cleaning)
import pandas as pd

nav_history = pd.read_csv("../data/raw/02_nav_history.csv")

nav_history.head()
```

```python
nav_history["date"] = pd.to_datetime(nav_history["date"])
nav_history["date"]
```

```python
nav_history.info()
```

```python
nav_history = nav_history.sort_values(["amfi_code", "date"])
nav_history
```

```python
nav_history = nav_history.drop_duplicates()
```

```python
invalid_nav = nav_history[nav_history["nav"] <= 0]

print(len(invalid_nav))
```

```python
nav_history["nav"] = (nav_history.groupby("amfi_code")["nav"].ffill())
nav_history["nav"]
```

```python
nav_history["nav"].isnull().sum()  #no null values
```

```python
# saves cleaned nav_history data to processed folder
nav_history.to_csv("../data/processed/nav_history_clean.csv",index=False)
```

```python
# designing sqllite star scheme 

import pandas as pd
from sqlalchemy import create_engine
```

```python
# Creating database

engine = create_engine("sqlite:///bluestock_mf.db")

print("Database Created")
```

```python
# loading cleaned data into database

fund_master = pd.read_csv("../data/raw/01_fund_master.csv")

fund_master.head()
```

```python
fund_master.to_sql("dim_fund",engine,
    if_exists="replace",index=False)

print("dim_fund loaded")
```

```python
# verifying 
pd.read_sql(
    "SELECT * FROM dim_fund LIMIT 5",
    engine
)
```

```python
# loading all cleaned datasets from processed folder.

nav_history = pd.read_csv("../data/processed/nav_history_clean.csv")

transactions = pd.read_csv("../data/processed/investor_transactions_clean.csv")

performance = pd.read_csv("../data/processed/scheme_performance_clean.csv")

aum = pd.read_csv("../data/raw/03_aum_by_fund_house.csv")
```

```python
nav_history.to_sql(
    "fact_nav",
    engine,
    if_exists="replace",
    index=False
)

print("fact_nav loaded")
```

```python
transactions.to_sql(
    "fact_transactions",
    engine,
    if_exists="replace",
    index=False
)

print("fact_transactions loaded")
```

```python
performance.to_sql(
    "fact_performance",
    engine,
    if_exists="replace",
    index=False
)

print("fact_performance loaded")
```

```python
aum.to_sql(
    "fact_aum",
    engine,
    if_exists="replace",
    index=False
)

print("fact_aum loaded")
```

```python
print("dim_fund:",len(fund_master))

print("fact_nav:",len(nav_history))

print("fact_transactions:",len(transactions))

print("fact_performance:",len(performance))

print("fact_aum:",len(aum))
```

```python
tables = {
    "dim_fund",
    "fact_nav",
    "fact_transactions",
    "fact_performance",
    "fact_aum"
}

for table in tables:

    query = f"SELECT COUNT(*) as count FROM {table}"

    count = pd.read_sql(
        query,
        engine
    )

    print(table)
    print(count)
    print("-"*40)
```
