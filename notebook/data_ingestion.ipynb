```python
import pandas as pd
import os

DATA_PATH = "../data/raw"

files = os.listdir(DATA_PATH)

files
```

```python
datasets = {}

for file in files:
    
    if file.endswith(".csv"):
        
        path = os.path.join(DATA_PATH, file)

        df = pd.read_csv(path)

        datasets[file] = df

print("Datasets Loaded:", len(datasets))
```

```python
master_codes = set(
    fund_master["amfi_code"]
)

nav_codes = set(
    nav_history["amfi_code"]
)

missing_codes = master_codes - nav_codes

print("Missing Codes:", len(missing_codes))
```

```python
# NOTE: 1. All AMFI codes in fund_master are present in nav_history.
#       2. Referential integrity is maintained.
```

```python
matched = len(master_codes.intersection(nav_codes))

total = len(master_codes)

print(
    "Match Percentage:",
    round((matched/total)*100,2),
    "%"
)
```

```python
import requests
import pandas as pd
```

```python
scheme_code = 125497

url = f"https://api.mfapi.in/mf/{scheme_code}"

response = requests.get(url)

print("Status Code:", response.status_code)
```

```python
data = response.json()

data.keys()
```

```python
data["meta"]
```

```python
nav_df = pd.DataFrame(data["data"])

nav_df.head()
```

```python
nav_df.columns
```

```python
nav_df.to_csv("../data/raw/hdfc_top100_live_nav.csv",index=False)

print("CSV Saved Successfully")
```

```python
schemes = {
    "SBI_Bluechip":119551,
    "ICICI_Bluechip":120503,
    "Nippon_LargeCap":118632,
    "Axis_Bluechip":119092,
    "Kotak_Bluechip":120841
}
```

```python
for name, code in schemes.items():

    url = f"https://api.mfapi.in/mf/{code}"
    response = requests.get(url)
    data = response.json()
    nav_df = pd.DataFrame(data["data"])
    nav_df.to_csv(f"../data/raw/{name}.csv",index=False)

    print(f"{name} downloaded successfully")
```
